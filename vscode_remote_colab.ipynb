# 💻 VS Code ↔ Google Colab 클라우드 런타임 원격 연동

이 노트북은 **Google Colab 클라우드 가상머신(GPU/CPU)**에 **VS Code Server(터널)**를 구동하여,
내 로컬 PC의 VS Code에서 코랩의 클라우드 자원을 그대로 사용할 수 있게 연결해주는 설정 스크립트입니다.

## 1. VS Code CLI 설치 및 터널(Tunnel) 시작
아래 셀을 실행하면 Microsoft 공식 VS Code 터널이 생성됩니다.  
처음 실행 시 GitHub 계정 로그인 인증 링크(8자리 코드)가 출력됩니다.

```python
# VS Code CLI 다운로드 및 터널 실행
!curl -Lk 'https://code.visualstudio.com/sha/download?build=stable&os=cli-alpine-x64' --output vscode_cli.tar.gz
!tar -xf vscode_cli.tar.gz

print("✨ 터널을 시작합니다. 터미널 출력에 나오는 GitHub 인증 링크(https://github.com/login/device)와 코드를 확인해 주세요!")
!./code tunnel --accept-server-license-terms
```

## 2. 내 로컬 VS Code에서 연결하기
1. 로컬 VS Code를 열고 **Remote - Tunnels** 확장을 설치합니다.
2. 좌측 하단 원격 아이콘(><) 클릭 -> **Connect to Tunnel** 선택.
3. 방금 위에서 인증한 동일한 GitHub 계정으로 로그인하면, 코랩 클라우드 인스턴스 이름이 목록에 나타납니다.
4. 연결 완료 시 로컬 VS Code에서 코랩 클라우드 터미널 및 파이썬 커널(Pandas 포함)을 그대로 사용할 수 있습니다!
